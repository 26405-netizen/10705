```python
import random
num=0
com=random.randint(1,10)
while num!=com:
  num=int(input('1~10 사이 수 입력: '))
  if num<com:
    print(f'{num}보다 큼')
  elif num>com:
    print(f'{num}보다 작음')
print('정답입니다')
```
